```python
#Decision Making
# Single statement suite
x = 10
if x > 5: print("x is greater than 5")
```

```python
#only if
x = 7
if x > 5:
    print("x is greater than 5")
```

```python
#if-else
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is 5 or less")
```

```python
#elif
x = 5
if x > 5:
    print("x is greater than 5")
elif x == 5:
    print("x is exactly 5")
else:
    print("x is less than 5")
```

```python
#nested if
x = 8
if x > 0:
    print("x is positive")
    if x % 2 == 0:
        print("x is even")
    else:
        print("x is odd")
else:
    print("x is zero or negative")
```

```python
#loops
# While loop example: Print numbers 1 to 5
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# For loop example: Iterate over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
#for with range

for i in range(10):
    print(i,end=' ')
print()
for i in range(1,10):
    print(i,end=' ')

print()
for i in range(0,10,3):
    print(i,end=" ")
```

```python
# Nested loop example: Multiplication table (1 to 3)
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i*j}")
    print("---")
```

```python
#Loop Control Statements
for num in range(1, 10):
    if num == 5:
        break  # Stop loop when num is 5
    print(num)
```

```python
for num in range(1, 6):
    if num == 3:
        continue  # Skip printing 3
    print(num)
```

```python
for num in range(1, 4):
    if num == 2:
        pass  # Placeholder for future code
    print(num)
```

```python
#iterator
# Creating an iterator from a list
numbers = [10, 20, 30]
it = iter(numbers)

print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
```

```python
# Generator function to yield squares
def square_numbers(n):
    for i in range(1, n+1):
        yield i * i

# Using the generator
for sq in square_numbers(5):
    print(sq)
```
