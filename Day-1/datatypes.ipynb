```python
# Python Lists Demonstration

# 1. Creating Lists
empty_list = []  # Empty list
numbers = [1, 2, 3, 4, 5]  # List of integers
mixed = [1, "Hello", 3.14, True]  # Mixed data types
nested = [[1, 2], [3, 4]]  # Nested list (matrix-like)

print("Created Lists:")
print(empty_list, numbers, mixed, nested)
```

```python
# 2. Accessing Values in Lists
print("\nAccessing Values:")
print("First element of numbers:", numbers[0])  # Index starts at 0
print("Last element of numbers:", numbers[-1])  # Negative index
print("Element from nested list:", nested[1][0])  # Accessing inside nested list
```

```python
# 3. Updating Lists
numbers[2] = 99  # Update element at index 2
print("\nUpdated numbers list:", numbers)
```

```python
# Adding elements
numbers.append(6)  # Add at end
numbers.insert(1, 50)  # Insert at index 1
print("After adding elements:", numbers)
```

```python
# 4. Deleting List Elements
del numbers[1]  # Delete by index
print("\nAfter deleting index 1:", numbers)

removed_value = numbers.pop()  # Remove last element
print("Popped value:", removed_value)
print("After pop:", numbers)

numbers.remove(99)  # Remove by value (first occurrence)
print("After removing 99:", numbers)
```

```python
# 5. Basic List Operators
list_a = [1, 2, 3]
list_b = [4, 5]
print("\nList Concatenation:", list_a + list_b)
print("List Repetition:", list_a * 2)
print("Membership Test:", 2 in list_a, 10 not in list_a)
```

```python
# 6. Indexing, Slicing, and Matrices
letters = ['a', 'b', 'c', 'd', 'e']
print("\nIndexing:", letters[2])
print("Slicing [1:4]:", letters[1:4])  # From index 1 to 3
print("Slicing with step [::2]:", letters[::2])  # Every second element

# Matrix (list of lists)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Matrix element [1][2]:", matrix[1][2])  # Row 1, Column 2
```

```python
# 7. Built-in List Functions & Methods
nums = [5, 2, 9, 1, 5, 6]
print("\nBuilt-in Functions:")
print("Length:", len(nums))
print("Max:", max(nums))
print("Min:", min(nums))
print("Sum:", sum(nums))
print("Sorted:", sorted(nums))  # Returns new sorted list

print("\nList Methods:")
nums.append(10)
print("Append:", nums)
nums.extend([20, 30])
print("Extend:", nums)
nums.sort()
print("Sort:", nums)
nums.reverse()
print("Reverse:", nums)
print("Count of 5:", nums.count(5))
nums.clear()
print("Clear:", nums)
```

```python
# Python Dictionary – Complete Example

# 1. Creating a Dictionary
# Using curly braces
student = {
    "name": "Alice",
    "age": 21,
    "courses": ["Math", "Physics"],
    "is_active": True
}

# Using dict() constructor
employee = dict(id=101, name="Bob", department="HR")

print("Student Dictionary:", student)
print("Employee Dictionary:", employee)
```

```python
# 2. Accessing Values in Dictionary
print("\nAccessing Values:")
print("Student Name:", student["name"])  # Direct access (KeyError if not found)
print("Student Age (safe):", student.get("age"))  # Safe access (returns None if not found)
print("Non-existing key with default:", student.get("grade", "Not Assigned"))
```

```python
# 3. Updating Dictionary
print("\nUpdating Dictionary:")
student["age"] = 22  # Update existing key
student["grade"] = "A"  # Add new key-value pair
student.update({"is_active": False, "city": "Chennai"})  # Update multiple keys
print("Updated Student:", student)
```

```python
# 4. Deleting Dictionary Elements
print("\nDeleting Elements:")
removed_value = student.pop("grade", None)  # Remove key and return value
print("Removed Grade:", removed_value)
del student["city"]  # Delete specific key
print("After Deletion:", student)
student.clear()  # Remove all items
print("After Clearing:", student)
```

```python
# 5. Properties of Dictionary Keys
# Keys must be immutable (string, number, tuple with immutable elements)
valid_dict = {
    1: "integer key",
    "key": "string key",
    (1, 2): "tuple key"
}
print("\nValid Keys Dictionary:", valid_dict)

# Invalid example (will raise TypeError if uncommented):
# invalid_dict = { [1, 2]: "list key" }  # Lists are mutable, so not allowed
```

```python
# 6. Built-in Dictionary Functions & Methods
sample = {"a": 1, "b": 2, "c": 3}

print("\nBuilt-in Functions & Methods:")
print("Length:", len(sample))
print("Keys:", list(sample.keys()))
print("Values:", list(sample.values()))
print("Items:", list(sample.items()))
print("Check if key exists:", "a" in sample)
print("Copy of dictionary:", sample.copy())

# fromkeys() – create new dict from keys with same value
new_dict = dict.fromkeys(["x", "y", "z"], 0)
print("From Keys:", new_dict)

# setdefault() – get value if key exists, else set default
print("Set Default:", sample.setdefault("d", 4))
print("After setdefault:", sample)

# popitem() – remove and return last inserted key-value pair
last_item = sample.popitem()
print("Popped Item:", last_item)
print("After popitem:", sample)
```

```python
# 1. Creating a Set
# Sets are unordered, unindexed, and do not allow duplicates

set_a = {1, 2, 3, 4}
set_b = set([3, 4, 5, 6])  # Using set() constructor
empty_set = set()  # Correct way to create an empty set ({} creates a dict)

print("Set A:", set_a)
print("Set B:", set_b)
print("Empty Set:", empty_set)
```

```python
# 2. Accessing a Set
# Sets are unordered, so indexing is not possible
# We can loop through elements

print("Accessing elements in Set A:")
for item in set_a:
    print(item)
```

```python
# 3. Modifying a Set
set_a.add(5)  # Add a single element
set_a.update([6, 7])  # Add multiple elements

print("After adding elements to Set A:", set_a)
```

```python
# 4. Removing Items from a Set
set_a.remove(7)  # Removes element; raises KeyError if not found
set_a.discard(10)  # Removes element if exists; no error if not found
popped_item = set_a.pop()  # Removes and returns an arbitrary element

print("After removals, Set A:", set_a)
print("Popped item:", popped_item)
```

```python
# 5. Common Set Operations
union_set = set_a | set_b  # Union
intersection_set = set_a & set_b  # Intersection
difference_set = set_a - set_b  # Difference
symmetric_diff_set = set_a ^ set_b  # Symmetric Difference

print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference (A - B):", difference_set)
print("Symmetric Difference:", symmetric_diff_set)
```

```python
# 6. Built-in Set Functions & Methods
print("Length of Set A:", len(set_a))
print("Is 3 in Set A?", 3 in set_a)
print("Is Set A subset of Set B?", set_a.issubset(set_b))
print("Is Set A superset of Set B?", set_a.issuperset(set_b))
print("Are Set A and Set B disjoint?", set_a.isdisjoint(set_b))

# Copying a set
set_c = set_a.copy()
print("Copied Set C:", set_c)

# Clearing a set
set_c.clear()
print("Cleared Set C:", set_c)
```

```python
#Pass by Reference vs Value
def modify_list(lst):
    lst.append(4)  # modifies original list

def modify_number(num):
    num += 10  # creates a new integer object

my_list = [1, 2, 3]
my_num = 5

modify_list(my_list)
modify_number(my_num)

print(my_list)  # [1, 2, 3, 4]
print(my_num)   # 5
```

```python
#required arguments
def add(a, b):
    return a + b

print(add(3, 4))  # 7
```

```python
#keyword arguments
def introduce(name, age):
    print(f"My name is {name} and I am {age} years old.")

introduce(age=25, name="Alice")
```

```python
#default args

def power(base, exponent=2):
    return base ** exponent

print(power(3))     # 9
print(power(3, 3))  # 27
```

```python
#variable length
def sum_all(*args):
    return sum(args)

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print(sum_all(1, 2, 3, 4))  # 10
print_info(name="Bob", age=30)
```
