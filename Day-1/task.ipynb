## Task 1 (Easy): Arithmetic Operators

Write a program that takes two numbers, `a` and `b`, and prints the result of:
sum, difference, product, division, floor division, modulus, and power.

Example: `a = 15`, `b = 4`

```python
# Solution
a = 15
b = 4

print('Sum:', a + b)
print('Difference:', a - b)
print('Product:', a * b)
print('Division:', a / b)
print('Floor Division:', a // b)
print('Modulus:', a % b)
print('Power:', a ** b)
```

## Task 2 (Easy): String Manipulation

Given a sentence, write a program that:
1. Prints the sentence in uppercase and lowercase
2. Counts the number of vowels in it
3. Prints the sentence reversed (using slicing)

Example: `sentence = "Python is fun to learn"`

```python
# Solution
sentence = "Python is fun to learn"

print('Uppercase:', sentence.upper())
print('Lowercase:', sentence.lower())

vowels = 'aeiouAEIOU'
vowel_count = 0
for ch in sentence:
    if ch in vowels:
        vowel_count += 1
print('Vowel count:', vowel_count)

print('Reversed:', sentence[::-1])
```

## Task 3 (Medium): Lists, Dictionaries and Loops

Given a list of numbers, write a program that:
1. Uses a `for` loop and `if` condition to split the numbers into a dictionary
   `{"even": [...], "odd": [...]}`
2. Prints the count, maximum, minimum, and sum of the original list

Example: `numbers = [12, 7, 5, 20, 3, 18, 9, 30, 1]`

```python
# Solution
numbers = [12, 7, 5, 20, 3, 18, 9, 30, 1]

result = {"even": [], "odd": []}
for n in numbers:
    if n % 2 == 0:
        result["even"].append(n)
    else:
        result["odd"].append(n)

print('Split by parity:', result)
print('Count:', len(numbers))
print('Max:', max(numbers))
print('Min:', min(numbers))
print('Sum:', sum(numbers))
```

## Task 4 (Medium): Functions, Generators and Loop Control

Write a generator function `classify(n)` that yields, for every number from 1 to `n`:
- `"Fizz"` if divisible by 3
- `"Buzz"` if divisible by 5
- `"FizzBuzz"` if divisible by both
- the number itself (as a string) otherwise

Then consume the generator with a `for` loop that:
- skips (using `continue`) any plain number (not Fizz/Buzz/FizzBuzz)
- stops (using `break`) as soon as 3 `"FizzBuzz"` results have been printed

Example: `n = 50`

```python
# Solution
def classify(n):
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            yield "FizzBuzz"
        elif i % 3 == 0:
            yield "Fizz"
        elif i % 5 == 0:
            yield "Buzz"
        else:
            yield str(i)

fizzbuzz_count = 0
for value in classify(50):
    if value not in ("Fizz", "Buzz", "FizzBuzz"):
        continue
    print(value)
    if value == "FizzBuzz":
        fizzbuzz_count += 1
        if fizzbuzz_count == 3:
            break
```
