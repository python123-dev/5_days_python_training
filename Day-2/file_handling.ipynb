# File Handling in Python

Covers the remaining Day-2 file topics from `Python_topics.md`:

- Reading and Writing Files
- File Positions
- File and Directory Operations
- Directories in Python

All examples work inside a temporary demo folder (created with `tempfile`) so running this
notebook does not leave any extra files behind in the project — everything is cleaned up in the
last cell.

```python
import os
import tempfile

# Create a temporary working directory for all the examples below
demo_dir = tempfile.mkdtemp(prefix="day2_file_handling_")
print("Working directory:", demo_dir)
```

## 1. Writing to a File

Use `open(path, mode)` with mode `'w'` (write, overwrites existing content) inside a `with`
block — the `with` statement automatically closes the file, even if an error occurs.

```python
file_path = os.path.join(demo_dir, "notes.txt")

with open(file_path, "w") as f:
    f.write("Line 1: Hello, file handling!\n")
    f.write("Line 2: Python makes this easy.\n")
    f.writelines(["Line 3: writelines example\n", "Line 4: last line\n"])

print("File written to:", file_path)
```

## 2. Reading from a File

Different ways to read: `read()` (whole file as one string), `readline()` (one line at a time),
`readlines()` (list of all lines), and iterating over the file object directly.

```python
# read(): entire file content as a single string
with open(file_path, "r") as f:
    content = f.read()
print("--- read() ---")
print(content)

# readline(): one line at a time
with open(file_path, "r") as f:
    print("--- readline() ---")
    print(f.readline(), end="")
    print(f.readline(), end="")

# readlines(): list of all lines
with open(file_path, "r") as f:
    lines = f.readlines()
print("--- readlines() ---")
print(lines)

# Iterating over the file object directly (memory-efficient, line by line)
print("--- iterating over file object ---")
with open(file_path, "r") as f:
    for line in f:
        print(line.strip())
```

## 3. Appending to a File

Mode `'a'` adds new content to the end of the file instead of overwriting it.

```python
with open(file_path, "a") as f:
    f.write("Line 5: appended later\n")

with open(file_path, "r") as f:
    print(f.read())
```

## 4. File Positions: `tell()` and `seek()`

`tell()` reports the current position (in bytes) of the file cursor. `seek(offset)` moves the
cursor to a specific position, letting you re-read or skip parts of a file.

```python
with open(file_path, "r") as f:
    first_chunk = f.read(6)
    print("Read so far:", repr(first_chunk))
    print("Current position (tell):", f.tell())

    # Move the cursor back to the beginning of the file
    f.seek(0)
    print("Position after seek(0):", f.tell())
    print("First line again:", f.readline().strip())

    # Skip ahead to a specific byte offset
    f.seek(7)
    print("After seek(7):", repr(f.read(5)))
```

## 5. File Operations: Exists, Rename, Copy, Size, Delete

The `os` and `os.path` modules provide functions to inspect and manipulate files.

```python
import shutil

print("File exists?", os.path.exists(file_path))
print("Is it a file?", os.path.isfile(file_path))
print("File size (bytes):", os.path.getsize(file_path))

# Copy the file
copy_path = os.path.join(demo_dir, "notes_copy.txt")
shutil.copy(file_path, copy_path)
print("Copy exists?", os.path.exists(copy_path))

# Rename the copy
renamed_path = os.path.join(demo_dir, "notes_renamed.txt")
os.rename(copy_path, renamed_path)
print("Renamed file exists?", os.path.exists(renamed_path))
print("Old copy name exists?", os.path.exists(copy_path))

# Delete the renamed copy
os.remove(renamed_path)
print("Renamed file exists after remove?", os.path.exists(renamed_path))
```

## 6. Directories in Python

Create, list, and remove directories using `os` — plus the modern `pathlib` alternative.

```python
# Create a subdirectory, and a nested one in a single call
sub_dir = os.path.join(demo_dir, "reports")
nested_dir = os.path.join(sub_dir, "2026", "august")

os.mkdir(sub_dir)                  # create a single directory
os.makedirs(nested_dir)            # create nested directories at once

print("sub_dir exists?", os.path.isdir(sub_dir))
print("nested_dir exists?", os.path.isdir(nested_dir))

# List contents of the working directory
print("Contents of demo_dir:", os.listdir(demo_dir))

# Current working directory
print("Current working directory:", os.getcwd())
```

```python
from pathlib import Path

# pathlib: an object-oriented alternative to os/os.path
p = Path(demo_dir)
print("Is directory:", p.is_dir())
print("Text files in demo_dir:", [f.name for f in p.glob("*.txt")])
print("All entries:", [entry.name for entry in p.iterdir()])

readme = p / "readme.md"
readme.write_text("# Demo\nCreated with pathlib.\n")
print("readme.md content:", readme.read_text())
```

## 7. Cleanup

Remove the entire temporary demo directory (and everything inside it) so the notebook leaves no
leftover files on disk.

```python
shutil.rmtree(demo_dir)
print("Cleaned up. demo_dir exists now?", os.path.exists(demo_dir))
```
