# Abstract Base Classes (ABC) in Python

An **Abstract Base Class** is a class that cannot be instantiated on its own — it exists to define
a common interface (a set of methods) that all its subclasses **must** implement.

- Defined using the built-in `abc` module: `ABC` (base class) and `@abstractmethod` (decorator).
- If a subclass does not implement every abstract method, it also cannot be instantiated.
- Useful for enforcing a common contract/interface across related classes (e.g. all `Shape`
  subclasses must have an `area()` method).

## Example 1: Basic Abstract Class

An abstract class cannot be instantiated directly, even though it looks like a normal class.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Trying to create an object of an abstract class raises a TypeError
try:
    s = Shape()
except TypeError as e:
    print("Error:", e)
```

## Example 2: Concrete Subclasses Implementing the Abstract Method

`Circle` and `Rectangle` both inherit from `Shape` and implement `area()`, so they can be
instantiated normally.

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

c = Circle(5)
r = Rectangle(4, 6)

print("Circle area:", c.area())
print("Rectangle area:", r.area())
```

## Example 3: Mixing Abstract and Concrete Methods

An abstract class can also define regular (concrete) methods that are shared by all subclasses.
Here `describe()` is a normal method that internally calls the abstract `area()` method — each
subclass supplies its own `area()`, but the `describe()` logic is written only once.

```python
class Shape2(ABC):
    @abstractmethod
    def area(self):
        pass

    # Concrete method, shared by every subclass
    def describe(self):
        print(f"This {self.__class__.__name__} has an area of {self.area():.2f}")

class Square(Shape2):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

sq = Square(4)
sq.describe()
```

## Example 4: Forgetting to Implement an Abstract Method

If a subclass does not override every abstract method from its parent, it is still considered
abstract and cannot be instantiated.

```python
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    # area() is NOT implemented here

try:
    t = Triangle(3, 4)
except TypeError as e:
    print("Error:", e)
```

## Example 5: Abstract Property

Besides abstract methods, you can also declare an abstract **property** by combining `@property`
with `@abstractmethod`. Subclasses must then provide their own property implementation.

```python
class Employee(ABC):
    @property
    @abstractmethod
    def salary(self):
        pass

class Manager(Employee):
    def __init__(self, base_salary, bonus):
        self.base_salary = base_salary
        self.bonus = bonus

    @property
    def salary(self):
        return self.base_salary + self.bonus

m = Manager(50000, 5000)
print("Manager salary:", m.salary)
```

## Example 6: `isinstance`/`issubclass` Checks with ABCs

ABCs work naturally with `isinstance()` and `issubclass()`, which is useful for verifying that a
class fulfils an expected interface before using it.

```python
print("issubclass(Circle, Shape):", issubclass(Circle, Shape))
print("isinstance(c, Shape):", isinstance(c, Shape))
print("isinstance(r, Circle):", isinstance(r, Circle))
```
