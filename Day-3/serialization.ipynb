```python
#JSON Serialization
import json

customer = {'id': 1, 'name': 'Ben', 'address': 'Park Lane 38', 'active': True}

json_string = json.dumps(customer, indent=2)
print(json_string)
print(type(json_string))
```

```python
#json.loads() - turn a JSON string back into a Python object
data = json.loads(json_string)
print(data)
print(type(data))
```

```python
#json.dump() / json.load() - same thing, but to/from a file
with open('customer.json', 'w') as f:
    json.dump(customer, f, indent=2)

with open('customer.json', 'r') as f:
    loaded = json.load(f)

print(loaded)
```

```python
#Pickle Serialization - can serialize almost any Python object, not just JSON-safe types
import pickle
from datetime import datetime

customer['joined'] = datetime.now()   #datetime isn't JSON serializable, but pickle handles it fine
print(customer)

pickled_bytes = pickle.dumps(customer)
print(pickled_bytes[:40], '...')
```

```python
#pickle.loads() - restore the original object, datetime and all
restored = pickle.loads(pickled_bytes)
print(restored)
print(type(restored['joined']))
```

```python
#pickle.dump() / pickle.load() - same thing, but to/from a file
with open('customer.pkl', 'wb') as f:
    pickle.dump(customer, f)

with open('customer.pkl', 'rb') as f:
    loaded_pickle = pickle.load(f)

print(loaded_pickle)
```
