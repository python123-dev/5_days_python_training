```python
#HTTP & HTTPS -> HyperText Transfer Protocol (Secure)
#HTTP runs on port 80 (plain text), HTTPS runs on port 443 (encrypted with SSL/TLS)
#requests handles both the same way, just change the scheme in the url
import requests

r = requests.get('https://www.python.org')
print(r.status_code)
print(r.url)
```

```python
#HTTP Methods
#GET->read data, POST->create, PUT->replace/update, PATCH->partial update, DELETE->remove
#postman-echo.com just echoes back whatever we send, useful for practice
```

```python
#GET request
response = requests.get('https://postman-echo.com/get', params={'name': 'gautham', 'course': 'python'})
print(response.status_code)
print(response.url)
print(response.json())
```

```python
#POST request
data = {'username': 'gautham', 'course': 'python'}
response = requests.post('https://postman-echo.com/post', json=data)
print(response.status_code)
print(response.json())
```

```python
#PUT request
response = requests.put('https://postman-echo.com/put', json={'status': 'updated'})
print(response.status_code)
print(response.json())
```

```python
#DELETE request
response = requests.delete('https://postman-echo.com/delete')
print(response.status_code)
print(response.json())
```

```python
#HTTP Requests and Responses
#a response object carries status_code, headers, text, content and json()
response = requests.get('https://postman-echo.com/get')
print(response.status_code)
print(response.headers['content-type'])
print(type(response.text))
print(type(response.content))
```

```python
#HTTP Headers - sending custom headers
headers = {'X-Custom-Header': 'TrainingDay3', 'User-Agent': 'python-training-client'}
response = requests.get('https://postman-echo.com/headers', headers=headers)
print(response.json())
```

```python
#HTTP Headers - reading response headers
response = requests.get('https://postman-echo.com/get')
for key, value in response.headers.items():
    print(key, ':', value)
```

```python
#Custom HTTP Requests - requests.request() lets you pick the method as a string
response = requests.request('GET', 'https://postman-echo.com/get', params={'q': 'custom'})
print(response.status_code)
print(response.json())
```

```python
#Status Codes
print(requests.get('https://postman-echo.com/status/200').status_code)   #success
print(requests.get('https://postman-echo.com/status/404').status_code)   #client error
print(requests.get('https://postman-echo.com/status/500').status_code)   #server error
```

```python
#raise_for_status() raises an exception for 4xx/5xx responses
try:
    r = requests.get('https://postman-echo.com/status/404')
    r.raise_for_status()
except requests.exceptions.HTTPError as e:
    print('Error:', e)
```

```python
#HTTP Authentication - basic auth with correct credentials
r = requests.get('https://postman-echo.com/basic-auth', auth=('postman', 'password'))
print(r.status_code)
print(r.json())
```

```python
#HTTP Authentication - basic auth with wrong credentials
r = requests.get('https://postman-echo.com/basic-auth', auth=('wrong', 'creds'))
print(r.status_code)
print(r.text)
```

```python
#HTTP Data Download - stream a file to disk in chunks
url = 'https://www.python.org/static/img/python-logo.png'
response = requests.get(url, stream=True)

with open('python_logo.png', 'wb') as f:
    for chunk in response.iter_content(chunk_size=1024):
        f.write(chunk)

print('downloaded, status:', response.status_code)
```
