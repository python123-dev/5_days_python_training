## 1. Introduction to Pandas

Where a NumPy array holds a single dtype in a grid, a pandas `DataFrame` behaves like a
spreadsheet or SQL table: labeled columns (which can each hold a different dtype) and a labeled
row index, with built-in tools for reading, cleaning, filtering, and summarizing data.

```python
# Solution
import pandas as pd
import numpy as np

print("pandas version:", pd.__version__)

# The sample dataset reused throughout this notebook: customer orders
orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "customer": ["Alice", "Bob", "Charlie", "Alice", "Diana", "Bob", "Eve", "Alice"],
    "category": ["Electronics", "Grocery", "Electronics", "Grocery", "Clothing", "Electronics", "Grocery", "Clothing"],
    "amount": [250.0, 45.5, 899.99, 60.0, 120.0, 300.0, np.nan, 75.0],
    "quantity": [1, 5, 1, 6, 2, 1, 3, 2],
})
print(orders)
```

## 2. Data Structures: `Series` & `DataFrame`

A `Series` is a single labeled column of data. A `DataFrame` is a dict-like collection of
`Series` that share the same row index. Both can be built from Python lists/dicts and given a
custom index.

```python
# Solution
prices = pd.Series([250.0, 45.5, 899.99, 60.0], index=["e1", "e2", "e3", "e4"], name="price")
print("Series:\n", prices)
print("Series index:", prices.index.tolist())
print("Series values:", prices.values)

# A DataFrame is a dict of Series sharing an index
df_from_dict = pd.DataFrame({
    "price": prices,
    "in_stock": [True, True, False, True],
})
print("\nDataFrame built from a dict of columns:\n", df_from_dict)
```

## 3. DataFrame Basic Properties

`shape`, `columns`, `index`, and `dtypes` describe a DataFrame's structure; `head()`/`tail()`
preview rows, and `info()` summarizes columns, dtypes, and non-null counts in one call.

```python
# Solution
print("shape:", orders.shape)
print("columns:", orders.columns.tolist())
print("index:", orders.index.tolist())
print("dtypes:\n", orders.dtypes)

print("\nhead(3):\n", orders.head(3))
print("\ntail(2):\n", orders.tail(2))

print()
orders.info()
```

## 4. Data Import & Export: CSV, Excel, HTML, JSON

Pandas can round-trip a DataFrame through several file formats: `to_csv`/`read_csv`,
`to_excel`/`read_excel` (needs `openpyxl`), `to_html`/`read_html` (needs `lxml`), and
`to_json`/`read_json`. A temporary directory keeps the notebook from leaving files behind.

```python
# Solution
import os
import tempfile

demo_dir = tempfile.mkdtemp(prefix="day4_pandas_")
small = orders.head(3)[["order_id", "customer", "amount"]]

# CSV
csv_path = os.path.join(demo_dir, "orders.csv")
small.to_csv(csv_path, index=False)
print("From CSV:\n", pd.read_csv(csv_path))

# Excel
xlsx_path = os.path.join(demo_dir, "orders.xlsx")
small.to_excel(xlsx_path, index=False)
print("\nFrom Excel:\n", pd.read_excel(xlsx_path))

# HTML
html_path = os.path.join(demo_dir, "orders.html")
small.to_html(html_path, index=False)
print("\nFrom HTML:\n", pd.read_html(html_path)[0])

# JSON
json_path = os.path.join(demo_dir, "orders.json")
small.to_json(json_path, orient="records")
print("\nFrom JSON:\n", pd.read_json(json_path))

for path in (csv_path, xlsx_path, html_path, json_path):
    os.remove(path)
os.rmdir(demo_dir)
print("\nCleaned up temporary files and directory")
```

## 5. Data Selection & Filtering

Select a column with `df['col']`, several columns with a list, rows by label with `.loc[]`, rows
by position with `.iloc[]`, and rows matching a condition with boolean indexing.

```python
# Solution
print("single column:\n", orders["customer"])
print("\nmultiple columns:\n", orders[["customer", "amount"]])
print("\n.loc rows 0-2, selected columns:\n", orders.loc[0:2, ["customer", "amount"]])
print("\n.iloc first 2 rows, first 3 columns:\n", orders.iloc[0:2, 0:3])

print("\nfiltered: amount > 100:\n", orders[orders["amount"] > 100])
print("\nfiltered: category == 'Electronics' AND quantity == 1:\n",
      orders[(orders["category"] == "Electronics") & (orders["quantity"] == 1)])
```

## 6. Descriptive Analysis

`.describe()` gives count/mean/std/min/quartiles/max for numeric columns in one call; individual
statistics (`mean`, `median`, `std`, `corr`) are also available directly.

```python
# Solution
print(orders.describe())
print("\nmean amount:", orders["amount"].mean())
print("median amount:", orders["amount"].median())
print("std amount:", orders["amount"].std())
print("\ncorrelation between amount and quantity:\n", orders[["amount", "quantity"]].corr())
```

## 7. Data Cleaning: Missing Values

`isna()` flags missing values, `dropna()` removes rows/columns containing them, and `fillna()`
replaces them (e.g. with the column mean).

```python
# Solution
print("missing values per column:\n", orders.isna().sum())
print("\nrows with a missing value:\n", orders[orders.isna().any(axis=1)])

dropped = orders.dropna()
print("\nafter dropna(), shape:", dropped.shape)

filled = orders.copy()
filled["amount"] = filled["amount"].fillna(filled["amount"].mean())
print("\nafter filling missing amount with the column mean:\n", filled)
```

## 8. Handling Unwanted Columns & Duplicate Entries

`drop(columns=...)` removes columns that aren't needed; `duplicated()`/`drop_duplicates()` find
and remove repeated rows.

```python
# Solution
with_dupe = pd.concat([orders, orders.iloc[[0]]], ignore_index=True)
print("shape with an injected duplicate row:", with_dupe.shape)
print("duplicated() flags:\n", with_dupe.duplicated())

deduped = with_dupe.drop_duplicates()
print("\nafter drop_duplicates(), shape:", deduped.shape)

trimmed = orders.drop(columns=["quantity"])
print("\nafter dropping the 'quantity' column:\n", trimmed.head())
```

## 9. Outlier Handling & Unique Values

A common outlier rule uses the interquartile range (IQR): flag values below `Q1 - 1.5*IQR` or
above `Q3 + 1.5*IQR`. `unique()`, `nunique()`, and `value_counts()` summarize distinct values in
a column.

```python
# Solution
amounts = filled["amount"]
q1, q3 = amounts.quantile(0.25), amounts.quantile(0.75)
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
print(f"Q1={q1}, Q3={q3}, IQR={iqr}, bounds=({lower}, {upper})")

outliers = filled[(amounts < lower) | (amounts > upper)]
print("outlier rows:\n", outliers)

no_outliers = filled[(amounts >= lower) & (amounts <= upper)]
print("\nshape without outliers:", no_outliers.shape)

print("\nunique categories:", orders["category"].unique())
print("number of unique categories:", orders["category"].nunique())
print("value counts per category:\n", orders["category"].value_counts())
```

## 10. Feature Creation from a Continuous Variable

`pd.cut()` buckets a continuous numeric column into labeled categorical bins — a common feature
engineering step (e.g. turning `amount` into a `spend_tier`).

```python
# Solution
tiered = filled.copy()
tiered["spend_tier"] = pd.cut(
    tiered["amount"],
    bins=[0, 75, 200, float("inf")],
    labels=["low", "medium", "high"],
)
print(tiered[["order_id", "amount", "spend_tier"]])
print("\ncount per tier:\n", tiered["spend_tier"].value_counts())
```

## 11. GroupBy Operations & Statistical Analysis

`groupby()` splits rows into groups by one or more columns; `.agg()` then applies one or more
statistics to each group — the "split-apply-combine" pattern.

```python
# Solution
by_category = orders.groupby("category")["amount"].sum()
print("total amount per category:\n", by_category)

by_customer = orders.groupby("customer").agg(
    total_spent=("amount", "sum"),
    orders_count=("order_id", "count"),
    avg_quantity=("quantity", "mean"),
)
print("\nper-customer summary:\n", by_customer)
```

## 12. The `apply` Method

`apply()` runs a custom function across a Series (element-wise) or a DataFrame (row-wise/
column-wise with `axis=1`/`axis=0`), for logic that doesn't fit a built-in method.

```python
# Solution
def classify(amount):
    if pd.isna(amount):
        return "unknown"
    return "big" if amount >= 200 else "small"

orders_with_flag = orders.copy()
orders_with_flag["size_flag"] = orders_with_flag["amount"].apply(classify)
print(orders_with_flag[["order_id", "amount", "size_flag"]])

# Row-wise apply: combine two columns into a derived value
orders_with_flag["unit_price"] = orders_with_flag.apply(
    lambda row: row["amount"] / row["quantity"] if row["quantity"] else np.nan,
    axis=1,
)
print("\nwith a row-wise derived 'unit_price' column:\n",
      orders_with_flag[["order_id", "amount", "quantity", "unit_price"]])
```
