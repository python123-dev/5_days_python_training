```python
import pandas as pd
```

```python
#### Reading from a csv file
df = pd.read_csv('https://raw.githubusercontent.com/pracasrv/machine-learning-getting-started-ws/master/data/Sample_Data.csv')
print(df)
print("#############################")
# df-data frame, df-2d matrix(m x n), m-number of rows, n- number of columns
# rows start from 0 index
# columns start from 0 index
```

#### Navigating through a Dataframe

```python
# get the records numbering 4,5,6
# : -> operator means range operator, 4:7 means 4,5,6(7 not included)
print(df[4:7])
print(df.iloc[4:7])
print("#############################")
```

```python
#### get the country value for the 5th record
# Index based access : access particular value based on m x n indexes(row x column) indexes, here 4 - row, 0 - country column
print(df.iloc[4,0])
print(df.loc[4,'Country'])

# accessing a particular value by directly specifiying the column name and row index
print(df['Country'][4])
print(df.Country[4])
```

```python
#### dropping columns
# drop(<<colum_name>>, axis=1)--- axis =1 refers to column
# drop(rowindex, axis=0)--- axis =0 refers to row
print(df)
print("#############################")
df1=df
df = df.drop('Country',axis=1)
print(df)
print("#############################")
```

```python
#### Adding Columns
df['Country'] = 'india'
#df['esal']=0
#df=df.drop('esal',1)
print(df)
```

```python
### dropping rows
# by using index, only one row can be dropped
# drop(rowindex, axis=0)--- axis =0 refers to row
print(df)
print("#############################")
df = df.drop(3,0)
print(df)
print("#############################")
```

```python
### Adding rows
df = pd.read_csv('https://raw.githubusercontent.com/pracasrv/machine-learning-getting-started-ws/master/data/Sample_Data.csv')
print(df)
print("#############################")
df2 = df[4:7]
df = df.drop(4,0)
df = df.drop(5,0)
df = df.drop(6,0)
print(df)
print("#############################")
print(df2)
```

```python
#append - add at the last
print(df2)
print("#############################")
df = df.append(df2)
print(df)
print("#############################")
```

```python
#### Sorting
print(df)
print("#############################")
df1 = df.sort_values(by='Age', ascending = False)
print(df1)
print("#############################")
```

```python
#### Boolean Indexing
df = pd.read_csv('https://raw.githubusercontent.com/pracasrv/machine-learning-getting-started-ws/master/data/Sample_Data.csv')
print(df)
print("#############################")
df.loc[df['Age']>40,'Salary'] = 0 #### Replace Salary as zero in rows where Age value is greater than 40
# df['Age']>40 - take all the rows whose age is greater than 40 and return the indexes(0,7,8,13,14)
# df.loc[0,7,8,13,14,'salary']=0 - assign the salary column to 0 for all rows with indexes(0,7,8,13,14)
# UPDATE customer SET Salary=0 WHERE age>40;
print(df)
print("#############################")
df.loc[((df['Age']>40) & (df['Country']=='Germany')),'Salary'] = 0 ##### Multiple Conditions
# UPDATE customer SET Salary=0 WHERE age>40 AND Country='Germary';
print(df)
print("#############################")
df.loc[((df['Age']>40) & (df['Country']=='France')),['Salary','Age']] = [9999,8888] ##### Multiple Conditions & multiple columns updates
# UPDATE customer SET Salary=0, Age=0 WHERE age>40 AND Country='France';
print(df)


```

```python
# dropping multiple rows using Conditions
df=df.drop(df.index[df.Salary == 0],axis = 0)
# DELETE FROM Customer WHERE salary=0;
print(df)
```
